Overall Sales Statistics

```python
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print(f"Total Sales       : ₹{total_sales:,.2f}")
print(f"Average Sales     : ₹{average_sales:,.2f}")
print(f"Highest Sale      : ₹{highest_sales:,.2f}")
print(f"Lowest Sale       : ₹{lowest_sales:,.2f}")
print(f"Total Quantity    : {total_quantity}")
print(f"Number of Orders  : {number_of_orders}")
```

Details of Highest-Sales Order

```python
highest_order = df.loc[df["Total_Sales"].idxmax()]

print(highest_order)
```

Details of Lowest-Sales Order

```python
lowest_order = df.loc[df["Total_Sales"].idxmin()]

print(lowest_order)
```

GroupBy – Category Analysis

```python
category_analysis = df.groupby("Category").agg(
    Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean")
)

category_analysis
```

Categories Sorted by Sales

```python
category_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)
```

GroupBy – City Analysis

```python
city_analysis = df.groupby("City").agg(
    Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean")
)

city_analysis
```

Cities Sorted by Total Sales

```python
city_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)
```

The top cities by total sales are:

```python
city_analysis.sort_values(
    by="Total_Sales",
    ascending=False
).head(5)
```

GroupBy – Product Analysis

```python
product_analysis = df.groupby("Product").agg(
    Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean")
)

product_analysis
```

Top 10 Products by Sales

```python
top_products = product_analysis.sort_values(
    by="Total_Sales",
    ascending=False
).head(10)

top_products
```

Product with Highest Average Sales

```python
product_analysis.sort_values(
    by="Average_Sales",
    ascending=False
).head(10)
```

GroupBy – Payment Method Analysis

```python
payment_analysis = df.groupby("Payment_Method").agg(
    Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean")
)

payment_analysis
```

Payment Methods Sorted by Sales

```python
payment_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)
```

Region-Wise Analysis

```python
region_analysis = df.groupby("Region").agg(
    Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean")
)

region_analysis.sort_values(
    by="Total_Sales",
    ascending=False
)
```

Find Orders Above Average Sales

```python
average_sales = df["Total_Sales"].mean()

above_average = df[
    df["Total_Sales"] > average_sales
]

print("Average Sales:", average_sales)
print("Orders above average:", len(above_average))

above_average[
    ["Order_ID", "Product", "Category", "City", "Total_Sales"]
].head(10)
```

Find Large Quantity Orders

```python
large_quantity_orders = df[df["Quantity"] > 3]

large_quantity_orders[
    ["Order_ID", "Product", "Quantity", "Total_Sales"]
]
```

Find Highly Rated Products/Orders

```python
high_rated = df[df["Rating"] >= 4.5]

high_rated[
    ["Order_ID", "Product", "Category", "Rating", "Total_Sales"]
]
```

Top 10 Individual Orders

Category and City Combined Analysis

```python
category_city = df.groupby(
    ["Category", "City"]
).agg(
    Orders=("Order_ID", "count"),
    Quantity=("Quantity", "sum"),
    Total_Sales=("Total_Sales", "sum")
).sort_values(
    by="Total_Sales",
    ascending=False
)

category_city.head(15)
```

Payment Method and Category Analysis

```python
payment_category = df.groupby(
    ["Payment_Method", "Category"]
).agg(
    Orders=("Order_ID", "count"),
    Total_Sales=("Total_Sales", "sum")
).sort_values(
    by="Total_Sales",
    ascending=False
)

payment_category
```
